```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("car.csv")
df
```

```python
df.describe()
```

```python
df1=df.drop(columns=['name'],axis=1)
```

```python
df1.info()
```

```python
df1.isna().sum()
```

```python
df1['km_driven']=df1['km_driven'].fillna(df1['km_driven'].mean())
```

```python
df1.isna().sum()
```

```python
df1=df1.dropna()
```

```python

df['owner'].unique()
```

```python
df['fuel'].value_counts()
```

```python
df['seller_type'].unique()
```

```python
df['transmission'].value_counts()
```

```python
df1['owner']=df1['owner'].replace({'First Owner':5, 'Second Owner':4, 'Fourth & Above Owner':3,
       'Third Owner':2, 'Test Drive Car':1})
df1['transmission']=df1['transmission'].replace({'Manual':1,'Automatic':2})
df1['fuel']=df1['fuel'].replace({'Diesel':5,'Petrol':4,'CNG':3,'LPG':2,'Electric':1})
df1['seller_type']=df1['seller_type'].replace({'Individual':3, 'Dealer':2, 'Trustmark Dealer':1})
```

```python
df1
```

```python
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()

df1[['year','km_driven','fuel','seller_type','transmission','owner']] = scaler.fit_transform(
    df1[['year','km_driven','fuel','seller_type','transmission','owner']]
)
```

```python
df1
```

```python
df1.corr()
```

```python
x1=df1[['transmission']]
y1=df1['selling_price']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
X_train,X_test,y_train,y_test=train_test_split(x1,y1, random_state=42, test_size=0.3)

print("X train=",X_train.shape)
print("Y train=",y_train.shape)
print("X test=",X_test.shape)
print("Y test=",y_test.shape)
```

```python
from sklearn.linear_model import LinearRegression

model1=LinearRegression().fit(X_train,y_train)
```

```python
y_pred=model1.predict(X_test)
y_pred

mse = mean_squared_error(y_test, y_pred)
mae = mean_absolute_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
print(f"R-squared: {r2}")
```

```python
user_trans = input("Enter the year of the car for prediction: ")
if user_trans.lower() == 'manual':
    user_trans1 = 1
elif user_trans.lower() == 'automatic':
    user_trans1 =2
user_input_df = pd.DataFrame({'transmission': [user_trans1]})


predicted_price = model1.predict(user_input_df)

# Access the first element of the array for formatting
print(f"For a car having transmission {user_trans}, the predicted selling price is: {predicted_price[0]:,.2f}")
```

```python
x2=df1.drop(columns=['selling_price'],axis=1)
y2=df1['selling_price']

X2_train,X2_test,y2_train,y2_test=train_test_split(x2,y2, random_state=42, test_size=0.3)

print("X train=",X2_train.shape)
print("Y train=",y2_train.shape)
print("X test=",X2_test.shape)
print("Y test=",y2_test.shape)
```

```python
X2_train
```

```python
model2=LinearRegression().fit(X2_train,y2_train)
```

```python
y_pred2=model2.predict(X2_test)
y_pred2

mse = mean_squared_error(y2_test, y_pred2)
mae = mean_absolute_error(y2_test, y_pred2)
r2 = r2_score(y2_test, y_pred2)

print(f"Mean Squared Error: {mse}")
print(f"Mean Absolute Error: {mae}")
print(f"R-squared: {r2}")
```

```python
owner_map = {
    'first owner': 5,
    'second owner': 4,
    'fourth & above owner': 3,
    'third owner': 2,
    'test drive car': 1
}

fuel_map = {
    'diesel': 5,
    'petrol': 4,
    'cng': 3,
    'lpg': 2,
    'electric': 1
}

seller_map = {
    'individual': 3,
    'dealer': 2,
    'trustmark dealer': 1
}

trans_map = {
    'manual': 1,
    'automatic': 2
}


owner_encoded = owner_map.get(input("Enter Owner Type: ").lower())
fuel_encoded = fuel_map.get(input("Enter Fuel Type: ").lower())
seller_encoded = seller_map.get(input("Enter Seller Type: ").lower())
trans_encoded = trans_map.get(input("Enter Transmission Type: ").lower())
km_driven = float(input("Enter KM Driven: "))
year=int(input("Enter Year: "))

user_input = pd.DataFrame({
    'year': [year],
    'km_driven': [km_driven],
    'fuel': [fuel_encoded],
    'seller_type': [seller_encoded],
    'transmission': [trans_encoded],
    'owner': [owner_encoded]
})

user_input = scaler.transform(user_input)

user_input_df = pd.DataFrame(user_input, columns=x2.columns)

predicted_price = model2.predict(user_input_df)

print(f"For the given car details, the predicted selling price is: {predicted_price[0]:,.2f}")
```
